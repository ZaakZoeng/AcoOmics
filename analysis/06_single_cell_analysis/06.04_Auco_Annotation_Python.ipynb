```python
import pandas as pd

# 1) 取 UMAP 坐标（n_obs × 2）
umap = adata_Auco.obsm["X_umap"]
umap_df = pd.DataFrame(umap, columns=["UMAP_1", "UMAP_2"], index=adata_Auco.obs_names)

# 2) 拼接需要的 metadata
meta_cols = ["Broad_cell_type", "Sub_cell_type", "seurat_clusters", "batch"]
meta_df = adata_Auco.obs[meta_cols].copy()

# 3) 合并并把细胞名作为一列导出
out_df = pd.concat([umap_df, meta_df], axis=1).reset_index().rename(columns={"index": "cell_name"})

# 4) 保存
out_df.to_csv(f"/share/home/zhangze/zz/NeuralOrigin/Data/06.SingleCellAnalysis/SingleCellAnnotation/Auco_UMAP_with_labels.csv", index=False)
```

```python
import json
import pandas as pd

# out_df 就是你之前导出的 dataframe（或重新从 csv 读入）
# out_df = pd.read_csv(f"{fig_dir}/Auco_UMAP_with_labels.csv")

# 组装 ECharts 数据点：value=[x,y] + 其它字段
data_points = []
for _, r in out_df.iterrows():
    data_points.append({
        "value": [float(r["UMAP_1"]), float(r["UMAP_2"])],
        "cell_name": r["cell_name"],
        "Broad_cell_type": r["Broad_cell_type"],
        "Sub_cell_type": r["Sub_cell_type"],
        "seurat_clusters": str(r["seurat_clusters"]),
        "batch": str(r["batch"]),
    })

# 组装一个可直接用于 option 的 json 片段（你前端也可以只取 data_points）
echarts_payload = {
    "xAxis": {},
    "yAxis": {},
    "series": [{
        "type": "scatter",
        "symbolSize": 3,      # UMAP 点通常很多，建议小一点；你可改 20
        "data": data_points
    }]
}

with open(f"{fig_dir}/22.Echarts_scatter_auco_UMAP.json", "w", encoding="utf-8") as f:
    json.dump(echarts_payload, f, ensure_ascii=False)
```
