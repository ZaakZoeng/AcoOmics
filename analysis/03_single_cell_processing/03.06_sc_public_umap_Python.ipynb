## 海葵 Neve

```python
adata_Neve = sc.read_h5ad("/share/home/zhangze/zz/NeuralOrigin/Data/03.SingleCellProcessing/SingleCellDataH5ad/Neve.normalized.h5ad")
adata_Neve
```

```python
print(f"Max value in adata.X: {adata_Neve.X.max()}")
print(f"Min value in adata.X: {adata_Neve.X.min()}")
```

```python
sc.pl.umap(adata_Neve, color=['CellType'])
```

```python
# ========== 基础绘图（不显示） ==========
sc.pl.umap(
    adata_Neve,
    color='CellType',
    legend_loc=None,
    show=False
)

# 获取当前图对象
fig = plt.gcf()
ax = plt.gca()

# ========== 第一部分：保存纯 UMAP 主图 ==========
# 去掉坐标轴、边框、图例
ax.set_title('')
ax.set_xlabel('')
ax.set_ylabel('')
ax.set_xticks([])
ax.set_yticks([])
ax.set_frame_on(False)

# 设置方形比例
ax.set_aspect('equal', adjustable='box')

# 保存纯净 UMAP
plt.savefig(figdir + "Neve.origin.umap_main.pdf",
            bbox_inches='tight', dpi=300)
            
plt.close(fig)

# ========== 第二部分：只绘制并保存 legend ==========
# 重新绘制带 legend 的 UMAP
sc.pl.umap(
    adata_Neve,
    color='CellType',
    show=False,
    legend_loc='right margin'   # 强制将 legend 绘制到右边
)

fig = plt.gcf()
ax = plt.gca()
ax.set_title('')

# 只保留 legend
for artist in ax.collections:
    artist.remove()

ax.set_axis_off()

plt.savefig(figdir + "Neve.origin.umap_legend.pdf",
            bbox_inches='tight', dpi=300)
plt.close(fig)
```

```python
# Neve：NV2.14439, NV2.14441, NV2.20231, NV2.20744, NV2.20991, NV2.21096
marker_genes = ['NV2.14439', 'NV2.14441', 'NV2.20231', 'NV2.20744', 'NV2.20991', 'NV2.21096']  # 替换为你感兴趣的基因
# 只保留adata中实际存在的基因
marker_genes_filtered = [g for g in marker_genes if g in adata_Neve.var_names]
print("实际可视化的marker基因:", marker_genes_filtered)
# 如果有筛选出的基因再画图，否则提醒
if marker_genes_filtered:
    sc.pl.umap(adata_Neve, color=marker_genes_filtered)
else:
    print("没有marker基因存在于adata.var_names中，无法绘图。")
```

```python
print(f"Max value in adata.X: {adata_Auco.X.max()}")
print(f"Min value in adata.X: {adata_Auco.X.min()}")
```

```python
# sc.pl.umap(adata_Auco, color=['seurat_clusters'])
# Scanpy 通常能正确处理，但显式转换可以避免一些潜在问题，特别是当聚类标签是纯数字时
if adata_Auco.obs['seurat_clusters'].dtype.name != 'category':
    adata_Auco.obs['seurat_clusters'] = pd.Categorical(adata_Auco.obs['seurat_clusters'])

adata_Auco.obs['seurat_clusters'] = 'Auco' + adata_Auco.obs['seurat_clusters'].astype(str)

sc.pl.umap(adata_Auco, color=['seurat_clusters'])

sc.pl.umap(
    adata_Auco,
    color=['seurat_clusters'],
    legend_loc='on data',      # 关键参数：在数据点上显示图例（即标签）
    legend_fontsize=10,        # 可选：调整标签字体大小
    legend_fontoutline=2,      # 可选：为标签文字添加轮廓，使其在背景中更突出
    add_outline=True,          # 可选：为 'on data' 的标签组添加轮廓线
    frameon=False              # 可选：移除图形边框，使图像更简洁
)
```
