```python
final_markers = ['OG0000036', 'OG0000112']

# Loop through each OG
for og in final_markers:
    print(f"Plotting marker: {og}")

    # -----------------------------
    # 1. Auco ST Violin
    # -----------------------------
    ax = sc.pl.violin(
        st_Auco_og,
        keys=[og],
        groupby="BroadType",
        jitter=0.4,
        rotation=45,
        stripplot=True,
        show=False
    )
    fig = ax.figure
    fig.savefig(f"/share/home/zhangze/zz/NeuralOrigin/Data/08.HubGenes/res/st_Hub_genes_violin_{og}.png", dpi=300, bbox_inches="tight")
    plt.close(fig)

    # -----------------------------
    # 2. Auco ST Spatial
    # -----------------------------
    fig, ax = plt.subplots(dpi=300)
    sc.pl.spatial(
        st_Auco_og,
        color=[og],
        vmin=0, vmax=1,
        spot_size=50,
        img=None,           # ⭐ 不显示组织图，只显示点
        legend_loc=None,    # ⭐ 不显示图例（可选）
        frameon=False,      # ⭐ 去掉边框
        # color_map='plasma',
        color_map='Reds',
        ax=ax,
        show=True
    )
    # ⭐ 设置透明背景
    ax.set_facecolor("none")  # 设置坐标轴背景为透明
    fig.patch.set_alpha(0)     # 设置图形背景透明度为0（完全透明）
    # 去掉坐标轴标签
    ax.set_xticks([])
    ax.set_yticks([])
    ax.set_xlabel("")
    ax.set_ylabel("")
    ax.set_title("")
    fig.savefig(f"/share/home/zhangze/zz/NeuralOrigin/Data/08.HubGenes/res/st_Hub_genes_spatial_{og}.png", dpi=300, bbox_inches="tight")
    plt.close(fig)

    # -----------------------------
    # 3. Auco SC Violin
    # -----------------------------
    ax = sc.pl.violin(
        sc_Auco_og,
        keys=[og],
        groupby="Broad_cell_type",
        jitter=0.4,
        rotation=45,
        stripplot=True,
        show=False
    )
    fig = ax.figure
    fig.savefig(f"/share/home/zhangze/zz/NeuralOrigin/Data/08.HubGenes/res/sc_Hub_genes_violin_{og}.png", dpi=300, bbox_inches="tight")
    plt.close(fig)

    # -----------------------------
    # 4. Auco SC UMAP
    # -----------------------------
    fig, ax = plt.subplots(dpi=300)

    sc.pl.umap(
        sc_Auco_og,
        color=[og],
        vmin=0, vmax=1,
        ax=ax,
        # color_map='plasma',
        color_map='Reds',
        show=True,
        legend_loc=None,   # ⭐ 不显示图例
        frameon=False      # ⭐ 去掉坐标轴边框
    )

    # ⭐ 去掉坐标刻度与标签、标题
    ax.set_xticks([])
    ax.set_yticks([])
    ax.set_xlabel("")
    ax.set_ylabel("")
    ax.set_title("")
    fig.savefig(f"/share/home/zhangze/zz/NeuralOrigin/Data/08.HubGenes/res/sc_Hub_genes_umap_{og}.png", dpi=300, bbox_inches="tight")
    plt.close(fig)

print("All individual OG marker plots saved (dpi=300).")
```
