# Placozoa TrH1

```python
import scanpy as sc
import pandas as pd

# 正确读取并转置表达矩阵
adata = sc.read_h5ad("/share/home/zhangze/zz/NeuralOrigin/Data/01.RawData/PublicData/2023Placozoa/Tadh/sc_Tadh.h5ad")
adata
```

```python
adata.var
```

```python
# 查找重复的基因名
duplicates = adata.var_names[adata.var_names.duplicated()]
# 输出重复的基因名
print("重复的基因名：", duplicates)
# 删除重复的基因（保留第一个出现的）
adata = adata[:, ~adata.var_names.duplicated()]
adata.var
```

```python
# 基因重命名，更换_为-
adata.var_names = adata.var_names.str.replace('_', '-', regex=False)
adata.var
```

```python
# 细胞名
adata.obs
```

```python
# 矩阵内容
# Values were log1p-normalized, mean-centered and scaled, and filtered for highly variable genes using the same procedure described above for downstream dimensionality reduction and visualization (e.g., PCA) using Scanpy. 
adata.X.toarray()
```

```python
max(adata.X.toarray()[0])
```

```python
adata
```

```python
# 标准化和 log 转换
sc.pp.normalize_total(adata, target_sum=1e4)    # 对应 scale.factor=10000
sc.pp.log1p(adata)
# 高度变异基因选择（如果需要）
sc.pp.highly_variable_genes(adata, n_top_genes=2000)
print(adata)
```

```python
# 保存原始数据
adata.raw = adata.copy()
```

```python
# 取高可变基因
adata = adata[:, adata.var.highly_variable]

# PCA
# sc.tl.pca(adata, svd_solver='arpack')
sc.tl.pca(
    adata,
    n_comps=50,          # 计算 50 个主成分（PCs）
    svd_solver='arpack',  # 默认方法
)

# 计算邻居图
# sc.pp.neighbors(adata)
sc.pp.neighbors(
    adata,
    n_neighbors=50,      # 明确指定 k=50
    n_pcs=50,            # 使用前 50 个 PCs
    method='umap',       # 默认基于 UMAP 的 KNN
    random_state=42      # 可复现性
)

# 计算 UMAP
# sc.tl.umap(adata)
sc.tl.louvain(
    adata,
    resolution=1.0,      # 明确指定分辨率=1
    random_state=42,     # 可复现性
    key_added='louvain'  # 聚类结果存储到 adata.obs['louvain']
)

# UMAP 可视化（基于 Louvain 聚类）
sc.tl.umap(
    adata,
    random_state=42,
    n_components=2           # 2D 可视化
)

# 绘制 UMAP 图（按 Louvain 聚类着色）
sc.pl.umap(
    adata,
    color="louvain",         # 使用 Louvain 聚类结果
    legend_loc="on data",    # 将簇标签显示在图上
    frameon=False,
    title="UMAP (Louvain clustering)"
)
```

```python
adata
```

```python
# 可视化 UMAP
sc.pl.umap(adata, color=['louvain'])
```

```python
# 可视化 UMAP
sc.pl.umap(adata, color=['cell_type'])
```

```python
# 可视化 UMAP
sc.pl.umap(adata, color=['broad_cell_type'])
```

```python
print(len(adata.obs["cell_type"].value_counts()))
adata.obs["cell_type"].value_counts()
```

```python
print(len(adata.obs["broad_cell_type"].value_counts()))
adata.obs["broad_cell_type"].value_counts()
```

```python
adata
```

```python
adata = adata.raw.to_adata()
adata
```

```python
# 保存adata数据
raw_TrH1_path = "/share/home/zhangze/zz/NeuralOrigin/Data/03.SingleCellProcessing/SingleCellDataH5ad/TrH1.normalized.h5ad"
adata.write(raw_TrH1_path)
```

```python
# 导出基因id
adata.var_names.to_series().to_csv('/share/home/zhangze/zz/NeuralOrigin/Data/03.SingleCellProcessing/SingleCellDataH5ad/TrH1.genes.txt', index=False, header=False)
```
