```python
import pandas as pd
import json

# 1. 设置路径
input_file = '/share/home/zhangze/zz/NeuralOrigin/Data/08.HubGenes/hub_genes_lists/NC-ECOGs_nodeCommon_12_15_enrichment_GO_results.csv'
output_json = '/share/home/zhangze/zz/NeuralOrigin/Data/08.HubGenes/hub_genes_lists/pathway_nce_cogs.json'

def convert_csv_to_pathway_json(csv_path, json_path):
    # 读取原始数据
    df = pd.read_csv(csv_path)

    # --- 执行您的数据清理逻辑 ---
    # 删除第一列 (Unnamed: 0)
    df = df.iloc[:, 1:]

    # 删除不需要的列
    cols_to_drop = ['GeneRatio', 'BgRatio', 'RichFactor', 'FoldEnrichment', 'zScore', 'qvalue']
    df = df.drop(columns=[c for c in cols_to_drop if c in df.columns])

    # 重命名 ID 为 GO_ID
    df = df.rename(columns={'ID': 'GO_ID'})

    # 显式保留并排序指定列
    final_columns = ['GO_ID', 'Description', 'pvalue', 'p.adjust', 'geneID', 'Count']
    df = df[final_columns]

    # --- 额外的优化处理 (为了前端展示更友好) ---
    # 填充缺失值
    df = df.fillna("-")
    
    # 科学计数法格式化 (保留三位有效数字)
    # 这样前端 el-table 排序时依然可以正确处理数值，同时展示美观
    df['pvalue'] = df['pvalue'].apply(lambda x: f"{x:.3e}" if x < 0.001 else round(x, 4))
    if 'p.adjust' in df.columns:
        df['p.adjust'] = df['p.adjust'].apply(lambda x: f"{x:.3e}" if x < 0.001 else round(x, 4))

    # --- 构建前端所需的 JSON 结构 ---
    result = {
        "count": len(df),
        "columns": df.columns.tolist(),
        "data": df.to_dict(orient='records') # 转换为 [{col1: val1, col2: val2}, ...]
    }

    # 导出文件
    with open(json_path, 'w', encoding='utf-8') as f:
        json.dump(result, f, ensure_ascii=False, indent=2)
    
    print(f"Successfully converted {len(df)} rows to {json_path}")

# 执行转换
convert_csv_to_pathway_json(input_file, output_json)
```

```python
import csv

# 1. 构建 term -> name 映射
term2name = {}
with open('/share/home/zhangze/zz/NeuralOrigin/Data/06.SingleCellAnalysis/eggNOG_ClusterProfiler_GO/term2name_GO.tsv', 'r', encoding='utf-8') as f:
    reader = csv.DictReader(f, delimiter='\t')
    for row in reader:
        term2name[row['term']] = row['name']

# 2. 读取 enrich_res.csv，只保留匹配的行，并替换 Description 列
input_file = '/share/home/zhangze/zz/NeuralOrigin/Data/06.SingleCellAnalysis/NeuronEmergence/Step3_gene_function/Auco_GOenrich_minpct0.3_logfc0.6.csv'
output_file = '/share/home/zhangze/zz/NeuralOrigin/Data/08.HubGenes/hub_genes_lists/NO-DEOGs_Auco_GOenrich_minpct0.3_logfc0.6.csv'

with open(input_file, 'r', encoding='utf-8') as infile, \
     open(output_file, 'w', encoding='utf-8', newline='') as outfile:

    reader = csv.DictReader(infile)
    fieldnames = reader.fieldnames
    writer = csv.DictWriter(outfile, fieldnames=fieldnames)
    writer.writeheader()

    for row in reader:
        term_id = row['Description']
        if term_id in term2name:          # 仅当能找到对应名称时才保留
            row['Description'] = term2name[term_id]
            writer.writerow(row)
        # 否则跳过该行（删除）

print(f"处理完成，结果已保存至 {output_file}（只保留匹配的行）")
```
