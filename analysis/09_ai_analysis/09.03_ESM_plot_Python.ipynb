**本文件说明：** 读取 ESM2 35M 的 OG0000036 距离矩阵，区别于 09.02 启用的 650M 分支。


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# ===== 1. 读取矩阵 =====
mat_file = "/share/home/zhangze/zz/NeuralOrigin/Data/09.AI/ESM/matrices/esm2_t12_35M/OG0000036.cosine_distance_matrix.tsv"
df = pd.read_csv(mat_file, sep="\t", index_col=0)

species_order = ["Spla", "ClH23", "HoH13", "TrH2", "TrH1", "Auco", "Clhe", "Neve", "Dare"]
df = df.loc[species_order, species_order]

# ===== 2. 绘图 =====
fig, ax = plt.subplots(figsize=(7.5, 6.5))
im = ax.imshow(df.values, cmap="viridis", vmin=0, vmax=0.08)

# 坐标轴
ax.set_xticks(np.arange(len(species_order)))
ax.set_yticks(np.arange(len(species_order)))
ax.set_xticklabels(species_order, rotation=90, fontsize=11)
ax.set_yticklabels(species_order, fontsize=11)

# 门类分隔线
for b in [0.5, 4.5, 7.5]:
    ax.axhline(b, color="white", lw=2)
    ax.axvline(b, color="white", lw=2)

# 标题
ax.set_title("OG0000036 cross-species ESM2 distance", fontsize=14)

# colorbar
cbar = plt.colorbar(im, ax=ax, fraction=0.046, pad=0.04)
cbar.set_label("Cosine distance", fontsize=12)

plt.tight_layout()
# plt.savefig("OG0000036.clean_heatmap.pdf")
# plt.savefig("OG0000036.clean_heatmap.png", dpi=300)
plt.show()
```
