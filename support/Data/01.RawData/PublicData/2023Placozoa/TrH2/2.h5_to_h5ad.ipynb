```python
# python-code for converting h5 to h5ad
import pandas as pd
import scanpy as sc


fp_mat = 'sc_TrH2.matrix.raw.h5'
fp_meta = 'sc_TrH2.metadata.csv'
adata_raw = sc.read_10x_h5(fp_mat)
metadata = pd.read_csv(fp_meta, index_col=0)
# add meta-data
for c in metadata.columns:
    if metadata[c].dtype == object:
        adata_raw.obs[c] = metadata[c].astype(str)  # 对象类型通常是字符串，明确转换以避免问题
    else:
        adata_raw.obs[c] = metadata[c]  # 直接赋值
print(adata_raw)
adata_raw.write('sc_TrH2.h5ad')
```
